```python

import os

DATA_PATH = '/content/drive/MyDrive/HIGGS_Decision_Tree'

print("Files in dataset folder:")

for file in os.listdir(DATA_PATH):
    print(file)
```

```python

import pandas as pd

# Load datasets
train_df = pd.read_parquet(DATA_PATH + '/train.parquet')
val_df = pd.read_parquet(DATA_PATH + '/validation.parquet')
test_df = pd.read_parquet(DATA_PATH + '/test.parquet')

print("Datasets loaded successfully!")
```

```python

print("Training Dataset Shape:", train_df.shape)
print("Validation Dataset Shape:", val_df.shape)
print("Testing Dataset Shape:", test_df.shape)
```

```python

print("Training Dataset Information:")
train_df.info()

print("\nFirst 5 Rows:")
display(train_df.head())
```

```python

print("TRAIN SHAPE:", train_df.shape)
print("VALIDATION SHAPE:", val_df.shape)
print("TEST SHAPE:", test_df.shape)

print("\nColumn names:")
print(train_df.columns.tolist())
```

```python

print("Missing values in Training:")
print(train_df.isnull().sum().sum())

print("\nMissing values in Validation:")
print(val_df.isnull().sum().sum())

print("\nMissing values in Testing:")
print(test_df.isnull().sum().sum())
```

```python

print("Training Target Distribution:")
print(train_df['target'].value_counts())

print("\nTarget Percentages:")
print(train_df['target'].value_counts(normalize=True) * 100)
```

```python
# Separate features (X) and target (y)

X_train = train_df.drop(columns=['target'])
y_train = train_df['target']

X_val = val_df.drop(columns=['target'])
y_val = val_df['target']

X_test = test_df.drop(columns=['target'])
y_test = test_df['target']

print("Training Features:", X_train.shape)
print("Training Target:", y_train.shape)

print("Validation Features:", X_val.shape)
print("Validation Target:", y_val.shape)

print("Testing Features:", X_test.shape)
print("Testing Target:", y_test.shape)
```

```python

from sklearn.model_selection import train_test_split

X_train_sample, _, y_train_sample, _ = train_test_split(
    X_train,
    y_train,
    train_size=0.10,
    random_state=42,
    stratify=y_train
)

print("Sample Training Shape:", X_train_sample.shape)
print("Sample Target Shape:", y_train_sample.shape)
```

```python

from sklearn.tree import DecisionTreeClassifier
import time

# Create baseline Decision Tree
baseline_model = DecisionTreeClassifier(
    random_state=42
)

# Train model
start_time = time.time()

baseline_model.fit(X_train_sample, y_train_sample)

training_time = time.time() - start_time

print("Baseline Decision Tree trained successfully!")
print("Training Time:", round(training_time, 2), "seconds")
```

```python

from sklearn.metrics import accuracy_score

# Training accuracy
train_pred = baseline_model.predict(X_train_sample)
train_accuracy = accuracy_score(y_train_sample, train_pred)

# Validation accuracy
val_pred = baseline_model.predict(X_val)
val_accuracy = accuracy_score(y_val, val_pred)

print("Training Accuracy:", round(train_accuracy * 100, 2), "%")
print("Validation Accuracy:", round(val_accuracy * 100, 2), "%")
```

```python

X_val_sample, _, y_val_sample, _ = train_test_split(
    X_val,
    y_val,
    train_size=0.20,
    random_state=42,
    stratify=y_val
)

print("Validation Sample:", X_val_sample.shape)
```

```python

import pandas as pd
import time

results_list = []

depth_values = [5, 10, 15, 20]
split_values = [2, 10, 50]

for depth in depth_values:
    for split in split_values:

        model = DecisionTreeClassifier(
            max_depth=depth,
            min_samples_split=split,
            random_state=42
        )

        start = time.time()

        model.fit(X_train_sample, y_train_sample)

        train_time = time.time() - start

        train_acc = model.score(X_train_sample, y_train_sample)
        val_acc = model.score(X_val_sample, y_val_sample)

        results_list.append({
            'max_depth': depth,
            'min_samples_split': split,
            'train_accuracy': train_acc,
            'validation_accuracy': val_acc,
            'overfitting_gap': train_acc - val_acc,
            'training_time': train_time
        })

        print(
            f"Depth: {depth}, Split: {split}, "
            f"Validation Accuracy: {val_acc:.4f}"
        )

tuning_df = pd.DataFrame(results_list)

display(
    tuning_df.sort_values(
        by='validation_accuracy',
        ascending=False
    )
)
```

```python

best_result = tuning_df.loc[
    tuning_df['validation_accuracy'].idxmax()
]

print("Best Hyperparameters:")
print(best_result)
```

```python

criteria_results = []

for criterion in ['gini', 'entropy']:

    model = DecisionTreeClassifier(
        criterion=criterion,
        max_depth=10,
        min_samples_split=50,
        random_state=42
    )

    start = time.time()

    model.fit(X_train_sample, y_train_sample)

    elapsed = time.time() - start

    train_acc = model.score(X_train_sample, y_train_sample)
    val_acc = model.score(X_val_sample, y_val_sample)

    criteria_results.append({
        'criterion': criterion,
        'train_accuracy': train_acc,
        'validation_accuracy': val_acc,
        'training_time': elapsed
    })

criteria_df = pd.DataFrame(criteria_results)

display(criteria_df)
```

```python
from sklearn.tree import DecisionTreeClassifier
import time

final_model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=10,
    min_samples_split=50,
    random_state=42
)

start_time = time.time()

final_model.fit(X_train_sample, y_train_sample)

final_training_time = time.time() - start_time

print("Final Model Training Completed!")
print("Training Time:", round(final_training_time, 2), "seconds")
print("Training Accuracy:", round(final_model.score(X_train_sample, y_train_sample) * 100, 2), "%")
print("Validation Accuracy:", round(final_model.score(X_val_sample, y_val_sample) * 100, 2), "%")
```

```python
from sklearn.tree import DecisionTreeClassifier
import time
import psutil

final_model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=10,
    min_samples_split=50,
    random_state=42
)

print("Starting Full Training...")

start_time = time.time()

final_model.fit(X_train, y_train)

final_training_time = time.time() - start_time

print("\nFull Training Completed!")

print("Training Time:", round(final_training_time, 2), "seconds")
print("Training Accuracy:", round(final_model.score(X_train, y_train) * 100, 2), "%")

print("Available RAM:",
      round(psutil.virtual_memory().available / (1024**3), 2), "GB")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Predict validation data
y_val_pred = final_model.predict(X_val)

# Calculate metrics
print("Validation Results")
print("-------------------------")

print("Accuracy:",
      round(accuracy_score(y_val, y_val_pred) * 100, 2), "%")

print("Precision:",
      round(precision_score(y_val, y_val_pred) * 100, 2), "%")

print("Recall:",
      round(recall_score(y_val, y_val_pred) * 100, 2), "%")

print("F1 Score:",
      round(f1_score(y_val, y_val_pred) * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_val, y_val_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['Background (0)', 'Signal (1)']
)

fig, ax = plt.subplots(figsize=(7, 6))
disp.plot(cmap='Blues', ax=ax, values_format=',d')

plt.title('Decision Tree - Confusion Matrix')
plt.show()

print("Confusion Matrix:")
print(cm)
```

```python
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

# Predict probabilities
y_val_prob = final_model.predict_proba(X_val)[:, 1]

# Calculate ROC and AUC
fpr, tpr, thresholds = roc_curve(y_val, y_val_prob)
roc_auc = roc_auc_score(y_val, y_val_prob)

# Plot ROC Curve
plt.figure(figsize=(8, 6))

plt.plot(
    fpr, tpr,
    label=f'Decision Tree (AUC = {roc_auc:.4f})'
)

plt.plot([0, 1], [0, 1], linestyle='--', label='Random Classifier')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Decision Tree - ROC Curve')

plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("ROC-AUC Score:", round(roc_auc, 4))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Get feature importance
feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': final_model.feature_importances_
})

# Sort by importance
feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

# Display top 15 features
print("Top 15 Important Features:")
display(feature_importance.head(15))

# Plot feature importance
plt.figure(figsize=(10, 7))

plt.barh(
    feature_importance.head(15)['Feature'][::-1],
    feature_importance.head(15)['Importance'][::-1]
)

plt.xlabel('Importance Score')
plt.ylabel('Features')
plt.title('Decision Tree - Top 15 Feature Importance')

plt.tight_layout()
plt.show()
```

```python
import os
import joblib

RESULTS_PATH = '/content/drive/MyDrive/HIGGS_Decision_Tree/results'

os.makedirs(RESULTS_PATH, exist_ok=True)

# Save feature importance
feature_importance.to_csv(
    RESULTS_PATH + '/feature_importance.csv',
    index=False
)

# Save confusion matrix
pd.DataFrame(
    cm,
    index=['Actual_0', 'Actual_1'],
    columns=['Predicted_0', 'Predicted_1']
).to_csv(RESULTS_PATH + '/confusion_matrix.csv')

# Save trained model
joblib.dump(
    final_model,
    RESULTS_PATH + '/decision_tree_model.joblib'
)

print("All results saved successfully!")
print("Saved location:", RESULTS_PATH)
```
